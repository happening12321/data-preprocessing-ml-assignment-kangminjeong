## 문제 2. 분석 가능한 고객 데이터로 정제

### 업무 상황

품질 진단 결과를 바탕으로 고객 데이터를 분석 가능한 상태로 정리해야 합니다.  
처리 과정에서 고객의 실제 구매 행동 정보가 불필요하게 손실되지 않도록 데이터 특성을 고려하세요.

### 요구사항

1. 원본 데이터를 보존한 상태에서 정제용 데이터를 생성하세요.
2. 분석에 맞지 않는 수치형·날짜형 자료형과 범주 표기를 정리하세요.
3. 결측치와 전체 행 기준 중복 데이터를 처리하세요.
4. 주요 수치형 컬럼의 이상치를 탐지하고 적절한 방법으로 처리하세요.
5. 처리 전후의 결측치, 중복값과 이상치 상태를 확인하세요.

### 힌트

- 변환할 수 없는 문자열은 결측치로 바꾼 뒤 일관되게 처리할 수 있습니다.
- 이상치는 무조건 삭제하기보다 실제 우수 고객의 행동일 가능성도 고려하세요.
- IQR은 이상치 후보를 확인하는 대표적인 방법입니다.

```python
# TODO: 원본을 보존하고 정제용 데이터 df_clean을 생성하세요.
df_clean = df.copy()

# TODO: 수치형·날짜형 자료형과 범주 표기를 정리하세요.

# 수치형
df_clean["Age"] = pd.to_numeric(df_clean["Age"], errors="coerce")
df_clean["AveragePurchaseAmount"] = pd.to_numeric(df_clean["AveragePurchaseAmount"], errors="coerce")

# 날짜형
df_clean["LastPurchaseDate"] = pd.to_datetime(df_clean["LastPurchaseDate"], errors="coerce")
df_clean["SignupDate"] = pd.to_datetime(df_clean["SignupDate"], errors="coerce")

# 범주형 표기 정리
df_clean["Gender"] = df_clean["Gender"].str.strip()
df_clean["Region"] = df_clean["Region"].str.strip()
df_clean["MembershipLevel"] = df_clean["MembershipLevel"].str.strip()
df_clean["CouponUsed"] = df_clean["CouponUsed"].str.strip()
df_clean["PreferredCategory"] = df_clean["PreferredCategory"].str.strip()
```

```python
# TODO: 결측치와 전체 행 기준 중복 데이터를 처리하세요.

# 결측치 처리
df_clean["Gender"] = df_clean["Gender"].fillna("Unknown")
df_clean["Age"] = df_clean["Age"].fillna(df_clean["Age"].median())
df_clean["Region"] = df_clean["Region"].fillna("Unknown")
df_clean["MembershipLevel"] = df_clean["MembershipLevel"].fillna("Unknown")
df_clean["AveragePurchaseAmount"] = df_clean["AveragePurchaseAmount"].fillna(
    df_clean["AveragePurchaseAmount"].median()
)
df_clean["SatisfactionScore"] = df_clean["SatisfactionScore"].fillna(
    df_clean["SatisfactionScore"].median()
)
df_clean["LastPurchaseDate"] = df_clean["LastPurchaseDate"].fillna(
    df_clean["LastPurchaseDate"].mode()[0]
)
df_clean["PreferredCategory"] = df_clean["PreferredCategory"].fillna("Unknown")

# 전체 행 기준 중복 제거
df_clean = df_clean.drop_duplicates()


# TODO: 처리 전후 결과를 확인하세요.

print("처리 후 결측치:")
print(df_clean.isnull().sum())

print("\n처리 후 중복 행:")
print(df_clean.duplicated().sum())

print("\n처리 전 행 수:", len(df))
print("처리 후 행 수:", len(df_clean))
```

```python
def get_iqr_bounds(series):
    # TODO: Q1, Q3, IQR, 하한과 상한을 반환하세요.
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    pass
    return Q1, Q3, IQR, lower, upper

# TODO: 주요 수치형 컬럼의 이상치를 탐지하고 처리하세요.
Q1_age, Q3_age, IQR_age, lower_age, upper_age = get_iqr_bounds(df_clean["Age"])
Q1_visit, Q3_visit, IQR_visit, lower_visit, upper_visit = get_iqr_bounds(df_clean["VisitCount"])
Q1_avg, Q3_avg, IQR_avg, lower_avg, upper_avg = get_iqr_bounds(df_clean["AveragePurchaseAmount"])
Q1_total, Q3_total, IQR_total, lower_total, upper_total = get_iqr_bounds(df_clean["TotalPurchaseAmount"])
Q1_score, Q3_score, IQR_score, lower_score, upper_score = get_iqr_bounds(df_clean["SatisfactionScore"])

print("처리 전 이상치")
print("Age:", ((df_clean["Age"] < lower_age) | (df_clean["Age"] > upper_age)).sum())
print("VisitCount:", ((df_clean["VisitCount"] < lower_visit) | (df_clean["VisitCount"] > upper_visit)).sum())
print("AveragePurchaseAmount:", ((df_clean["AveragePurchaseAmount"] < lower_avg) | (df_clean["AveragePurchaseAmount"] > upper_avg)).sum())
print("TotalPurchaseAmount:", ((df_clean["TotalPurchaseAmount"] < lower_total) | (df_clean["TotalPurchaseAmount"] > upper_total)).sum())
print("SatisfactionScore:", ((df_clean["SatisfactionScore"] < lower_score) | (df_clean["SatisfactionScore"] > upper_score)).sum())


# 이상치를 경계값으로 대체
df_clean["Age"] = df_clean["Age"].clip(lower_age, upper_age)
df_clean["VisitCount"] = df_clean["VisitCount"].clip(lower_visit, upper_visit)
df_clean["AveragePurchaseAmount"] = df_clean["AveragePurchaseAmount"].clip(lower_avg, upper_avg)
df_clean["TotalPurchaseAmount"] = df_clean["TotalPurchaseAmount"].clip(lower_total, upper_total)
df_clean["SatisfactionScore"] = df_clean["SatisfactionScore"].clip(lower_score, upper_score)

# TODO: 처리 전후 이상치 상태를 확인하세요.

print("\n처리 후 이상치")
print("Age:", ((df_clean["Age"] < lower_age) | (df_clean["Age"] > upper_age)).sum())
print("VisitCount:", ((df_clean["VisitCount"] < lower_visit) | (df_clean["VisitCount"] > upper_visit)).sum())
print("AveragePurchaseAmount:", ((df_clean["AveragePurchaseAmount"] < lower_avg) | (df_clean["AveragePurchaseAmount"] > upper_avg)).sum())
print("TotalPurchaseAmount:", ((df_clean["TotalPurchaseAmount"] < lower_total) | (df_clean["TotalPurchaseAmount"] > upper_total)).sum())
print("SatisfactionScore:", ((df_clean["SatisfactionScore"] < lower_score) | (df_clean["SatisfactionScore"] > upper_score)).sum())
```
